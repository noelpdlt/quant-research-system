# Analysis

```python
df = pd.read_csv("../outputs/experiment_results.csv")

plt.figure(figsize=(7,7))

plt.scatter(df["Train Sharpe"], df["Test Sharpe"])

lims = [
    min(df["Train Sharpe"].min(), df["Test Sharpe"].min()),
    max(df["Train Sharpe"].max(), df["Test Sharpe"].max())
]

plt.plot(lims, lims, "--")

for _, row in df.iterrows():
    plt.text(
        row["Train Sharpe"],
        row["Test Sharpe"],
        row["Asset"]
    )

plt.xlabel("Train Sharpe")
plt.ylabel("Test Sharpe")
plt.title("Train vs Test Sharpe")

plt.show()
```

```python
df["Decay"] = (
    df["Test Sharpe"] -
    df["Train Sharpe"]
)

decay = (
    df.groupby("Asset")["Decay"]
      .mean()
      .sort_values()
)

decay.plot.bar(figsize=(10,5))

plt.ylabel("Average Sharpe Decay")

plt.show()
```

```python
summary = (
    df.groupby("Asset")
      .agg({
          "Train Sharpe":"mean",
          "Test Sharpe":"mean",
          "Train Volatility":"mean",
          "Test Volatility":"mean",
          "Train Max DD":"mean",
          "Test Max DD":"mean"
      })
      .sort_values("Test Sharpe", ascending=False)
)

summary
```

```python
plt.figure(figsize=(8,5))

df.boxplot(
    column="Test Sharpe",
    by="Strategy",
    rot=30
)

plt.title("Test Sharpe by Strategy")
plt.suptitle("")
plt.ylabel("Sharpe")

plt.show()
```

```python
df["Sharpe Decay"] = (
    df["Test Sharpe"] -
    df["Train Sharpe"]
)

decay = (
    df.groupby("Strategy")["Sharpe Decay"]
      .mean()
)

decay.plot.bar(figsize=(8,5))

plt.ylabel("Average Test − Train Sharpe")

plt.show()
```

```python
summary = (
    df.groupby("Strategy")
      [["Train Sharpe","Test Sharpe"]]
      .mean()
)

summary.plot.bar(figsize=(8,5))

plt.ylabel("Average Sharpe")
plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.scatter(
    df["Window"],
    df["Test Sharpe"]
)

plt.xlabel("Optimized Window")
plt.ylabel("Test Sharpe")

plt.show()
```

```python
plt.hist(
    df["Test Sharpe"],
    bins=10
)

plt.xlabel("Test Sharpe")
plt.show()
```

```python
metrics = df[[
    "Train Sharpe",
    "Train Volatility",
    "Train Max DD",
    "Test Sharpe",
    "Test Volatility",
    "Test Max DD"
]]

metrics.corr()
```

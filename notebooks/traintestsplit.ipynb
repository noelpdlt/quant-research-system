```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

sys.path.append(str(Path().resolve().parent))

from src.data import get_data
from src.features import generate_features
from src.backtest import backtest, results_table, optimize, optimize_assets
from src.strategy import momentum_trend_strategy, momentum_volatility_strategy, momentum_voluratio_strategy, momentum_strategy, mean_reversion_strategy
from src.visualization import plot_performance
```

```python
windows = range(1, 51)
strategies = [
    ('Momentum', momentum_strategy, 'mom', None),
    ('Mean Reversion', mean_reversion_strategy, 'mr', range(2, 100, 2)),
    ("Momentum + Trend Strength", momentum_trend_strategy, "mo_tr", None),
    ("Momentum + Rolling Volatility", momentum_volatility_strategy, "mo_vo", range(2, 100, 2)),
    ("Momentum + Volume Ratio", momentum_voluratio_strategy, "mo_vr", None)
]

assets = ['AAPL', 'MSFT', 'SPY', 'NVDA']
```

```python
optimize_assets(
    assets=assets,
    start_date="2016-01-01",
    end_date="2026-01-01",
    strategies=strategies,
    train_ratio = 0.7,
    windows=range(2, 50, 2),    
)
```

## AAPL Train / Test

```python
base_df = get_data('AAPL', '2016-01-01', '2026-01-01')
base_df = generate_features(base_df)

train = base_df.loc["2016-01-01":"2021-12-31"]
test = base_df.loc["2022-01-01":"2025-12-31"]
```

```python
df_train = train.copy()
optimize(df_train, strategies, windows)
```

```python
df_test = test.copy()

df_test = momentum_strategy(df_test, 16)
df_test = mean_reversion_strategy(df_test, 4, 0.02)
df_test = momentum_trend_strategy(df_test, 10)
df_test = momentum_volatility_strategy(df_test, 14, 0.02)
df_test = momentum_voluratio_strategy(df_test, 21)

df_test = backtest(df_test)

plot_performance(df_test)
```

```python
results_table(df_test)
```
